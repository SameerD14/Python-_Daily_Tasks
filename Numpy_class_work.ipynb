```python
import pandas as pd
import numpy as np

np.random.seed(10)
id=np.arange(100,600)
name=np.random.choice(["Ravi","Amit","Rahul","Sameer","Vikas",
     "Priya"], 500)
gender=np.random.choice(["Male","Female"],500)
dept = np.random.choice(
    ["IT","HR","Finance","Sales","Marketing","Operations"],500)
exp = np.random.randint(0,16,500)
role = np.random.choice(
    ["Data Analyst","Data Scientist","Software Engineer",
     "HR Executive"], 500)
salary=np.random.randint(25000,120001,500)
df=pd.DataFrame({
    "id":id,
    "name":name,
    "gender":gender,
    "dept":dept,
    "exp":exp,
    "role":role,
    "salary":salary})
print(df)
print(df.shape)
```
