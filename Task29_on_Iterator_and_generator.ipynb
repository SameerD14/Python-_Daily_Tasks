##### Problem 6
##### Create a custom iterator that prints numbers from 1 to 10.

```python
class Count:
    def __init__(self):
        self.num=1
    def __iter__(self):
       return self
    def __next__(self):
        if self.num<=10:
            value=self.num
            self.num+=1
            return value
        else:
            raise StopIteration
obj=Count()
for i in obj:
    print(i)
```

##### Problem 7
##### Create a custom iterator that prints only even numbers from 2 to 20.

```python
class Count:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        while self.num <= 20:
            if self.num % 2 == 0:
                value = self.num
                self.num += 1
                return value
            self.num += 1

        raise StopIteration


obj = Count()

for i in obj:
    print(i)
```

##### Problem 20
##### Create a generator that yields the length of each word in the
##### following list.
##### words = ["Python", "Java", "AI", "Generator", "Iterator"]

```python
words = ["Python", "Java", "AI", "Generator", "Iterator"]
def length_of_words():
    for word in words:
        yield (word)
gen=length_of_words()
for i in gen:
    print(len(i))
```

##### Problem 23
##### Create a custom iterator that prints all alphabet letters from A to Z.

```python
class Alphabet():
    def __init__(self):
        self.char=ord("A")
    def __iter__(self):
        return self
    def __next__(self):
        if self.char<=ord("Z"):
            value=chr(self.char)
            self.char+=1
            return value
        else:
            raise StopIteration
obj=Alphabet()
for i in obj:
    print(i)
```

##### Problem 25
##### Create a custom iterator that prints the Fibonacci sequence up to 10
##### terms.

```python
class Fibonacci():
    def __init__(self):
        self.a=0
        self.b=1
        self.count=1
    def __iter__(self):
        return self
    def __next__(self):
        if self.count<=10:
            value=self.a
            self.a=self.b
            self.b=value+self.b
            self.count+=1
            return value
        else:
            raise StopIteration

obj=Fibonacci()
for i in obj:
    print(i)
```
