```python
# module 2
import pandas as pd
df=pd.read_csv("C:\\Users\\hp\\Downloads\\students2.csv")
df.head(15)
```

```python
# Display the first 5 rows using loc[].
print(df.loc[0:4])
print()
# Display the first 5 rows using iloc[].
print(df.loc[0:5])
# Display only the Name column using loc[].
print()
print(df.loc[0:5,"Name"])
# Display the Name and Marks columns using iloc[].
print()
print(df.iloc[0:5,[1,5]])
# Show students with marks greater than 85.
print()
print(df[df["Marks"]>85])
# Show students with marks less than 75.
print()
print(df[df["Marks"]<75])
# Show students whose age is 22.
print()
print(df[df["Age"]==22])
# Show students whose city is Hyderabad.
print()
print(df[df["City"]=="Hyderabad"])
# Show students whose city is not Pune.
print()
print(df[df["City"]!="Pune"])
# Show students whose marks are greater than or equal to 90.
print()
print(df[df["Marks"]>=90])
# Show students whose marks are between 80 and 90.
print()
print(df[(df["Marks"]>=80 )& (df["Marks"]<=90)])
# Show students whose age is between 21 and 22.
print()
print(df[df["Age"].between(20,21)])
# Show students from Pune or Mumbai.
print()
print(df[(df["City"]=="Pune") | (df["City"]=="Mumbai")])
# Show students from Hyderabad and marks greater than 80.
print()
print(df[(df["City"]=="Hyderabad")&(df["Marks"]>80)])
# Show students whose city is Pune, Mumbai, or Delhi using isin().
print()
print(df[df["City"].isin(["Pune","Mumbai","Delhi"])])
# Show only the names of students whose marks are greater than 85 using loc[].
print()
print(df.loc[df["Marks"]>85,"Name"])
# Find all female students with marks above 85.
print()
print(df.loc[(df["Marks"] > 85) & (df["Gender"] == "Female"), "Name"])# Find all male students from Delhi.
# Show rows 10 to 20 using loc[]
print()
print(df.loc[10:20])
# Show rows 10 to 20 and only the Name and City columns using iloc[].
print()
print(df.iloc[10:21])
```

```python
# Module 5
# Basic
# Sort students by Marks in ascending order.
print(df.sort_values(by="Marks"))
# Sort students by Marks in descending order.
print()
print(df.sort_values(by="Marks",ascending=False))
# Sort students by Age.
print()
print(df.sort_values(by="Age"))
# Sort students by Name alphabetically.
print()
print(df.sort_values(by="Name"))
# Sort students by City.
print()
print(df.sort_values(by="City"))
# Intermediate
# Sort students by City and then by Marks.
print()
print(df.sort_values(by=["City","Marks"]))
# Sort students by City (A–Z) and Marks (highest first).
print()
print(df.sort_values(by=["City","Marks"],ascending=[True,False]))
# Sort the DataFrame by index in ascending order.
print()
print(df.sort_index())
# Sort the DataFrame by index in descending order.
print()
print(df.sort_index(ascending=False))
# Advanced
# Show the top 5 students with the highest marks.
print()
print(df.nlargest(5,"Marks"))
# Show the top 3 oldest students.
print()
print(df.nlargest(3,"Age"))
# Show the 5 students with the lowest marks.
print()
print(df.nsmallest(5,"Marks"))
# Show the 10 youngest students.
print()
print(df.nsmallest(10,"Age"))
# Find the top 5 students from the sorted DataFrame by marks.
print()
sorted_df = df.sort_values(by="Marks", ascending=False)

print(sorted_df.head(5))

# Compare the output of sort_values(by="Marks") and nlargest(5, "Marks").
print()
print(df.sort_values(by="Marks", ascending=False))
print()
print(df.nlargest(5,"Marks"))
```
