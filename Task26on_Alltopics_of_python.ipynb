##### 14. Create class BankAccount with private attribute __balance. Add deposit and withdraw
##### methods with validation.

```python
class BankAccount:
    def __init__(self,acc_no,balance):
        self.acc_no=acc_no
        self.__balance=balance
    def get_balance(self):
        print("Balance=",self.__balance)
    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            print("Deposit amount=",amount)
        else:
            print("Invalid amount")
        return self.__balance
    def withdraw(self,amount):
        if amount<0:
            print("Invalid amount")
        elif amount>self.__balance:
            print("Insufficient balance")
        else :
            self.__balance-=amount
            print("Withdraw amount=",amount)
        return self.__balance
user1=BankAccount(89980,8000)
user1.get_balance()
user1.deposit(5000)
user1.get_balance()
user1.withdraw(7000)
user1.get_balance()
```

##### 18. Create class Rectangle with private length and width. Add getter, setter and method to calculate area.

```python
class Rectraingle:
    def __init__(self,length,width):
        self.__length=length
        self.__width=width
    def set_length(self,length):
        if length>0:
            self.__length=length
        else:
            print("Invalid length")
    def get_length(self):
        print("Length=",self.__length)
    def set_width(self,width):
        if width>0:
            self.__width=width
        else:
            print("Invalid length")
    def get_width(self):
        print("Width=",self.__width)
    def area(self):
        print("Area=",self.__length*self.__width)
rect=Rectraingle(8,4)
rect.get_length()
rect.get_width()
rect.area()
rect.set_length(9)
rect.set_width(5)
rect.get_length()
rect.get_width()
rect.area()
    
```

##### Inheritance + Abstraction + super() (19-24)
##### 19. Create single level inheritance: Person → Student. Add attributes in both classes.

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show_person(self):
        print("Name:",self.name)
        print("Age:",self.age)

class Student(Person):
    def __init__(self,name, age, roll_no):
        super().__init__(name,age)
        self.roll_no=roll_no

    def show_student(self):
        print("Roll no:",self.roll_no)

s1=Student("Sameer",22,101)
s1.show_person()
s1.show_student()
```

##### 20. Create multi-level inheritance: Person → Employee → Manager.

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show_person(self):
        print("Name:",self.name)
        print("Age:",self.age)

class Employee(Person):
    def __init__(self,name, age, emp_id):
        super().__init__(name,age)
        self.emp_id=emp_id

    def show_employee(self):
        print("Employee Id:",self.emp_id)

class Manager(Employee):
    def __init__(self,name, age, emp_id, department):
        super().__init__(name,age,emp_id)
        self.department=department

    def show_manager(self):
        print("Department:",self.department)

p1=Manager("Sanket",22, 11101, "IT")
p1.show_person()
p1.show_employee()
p1.show_manager()
```

##### 21. Create hierarchical inheritance: Parent Animal → Children Dog and Cat.

```python
class Animal:
    def show(self):
        print("Animals")
class Dog(Animal):
    def __init__(self,breed):
        self.breed=breed
    def display(self):
        self.show()
        print("Breed=",self.breed)
class Cat(Animal):
    def __init__(self,color):
        self.color=color
    def display(self):
        self.show()
        print("Color=",self.color)
d=Dog("labra")
d.display()
c=Cat("green")
c.display()
```
