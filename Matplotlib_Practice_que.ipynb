###### Q1. Basic Line Chart
###### Create a line chart for:
###### months = ["Jan", "Feb", "Mar", "Apr", "May"]
###### sales = [100, 150, 120, 180, 200]
###### Requirements:
###### Line chart
###### Marker on each point
###### Title: "Monthly Sales"
###### X-axis: "Month"
###### Y-axis: "Sales"

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales= [100, 150, 120, 180, 200]
plt.plot(months,sales)
plt.xlabel("Months")
plt.title("Monthly Sales")
plt.ylabel("sales")
plt.show()
```

###### Q7. Subplots Create a 2 × 2 subplot containing:
###### Line chart → Monthly Sales
###### Bar chart → Product Sales
###### Histogram → Marks
###### Scatter plot → Study Hours vs Marks
###### Add a main figure title:
###### Data Analysis Charts
###### Very important: Remember:
###### fig, axes = plt.subplots(2, 2) and the indexes:
###### axes[0,0]
###### axes[0,1]
###### axes[1,0]
###### axes[1,1]

```python
import matplotlib.pyplot as plt

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]

products = ["Laptop", "Mobile", "Tablet", "Watch"]
product_sales = [50, 80, 40, 30]

marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90]

study_hours = [1, 2, 3, 4, 5, 6]
student_marks = [40, 45, 55, 60, 70, 80]


# Create 2 × 2 subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 7))


# 1. Line Chart
axes[0,0].plot(months, sales, marker="o")
axes[0,0].set_title("Monthly Sales")
axes[0,0].set_xlabel("Month")
axes[0,0].set_ylabel("Sales")


# 2. Bar Chart
axes[0,1].bar(products, product_sales)
axes[0,1].set_title("Product Sales")
axes[0,1].set_xlabel("Product")
axes[0,1].set_ylabel("Sales")


# 3. Histogram
axes[1,0].hist(marks, bins=5)
axes[1,0].set_title("Marks Distribution")
axes[1,0].set_xlabel("Marks")
axes[1,0].set_ylabel("Frequency")


# 4. Scatter Plot
axes[1,1].scatter(study_hours, student_marks)
axes[1,1].set_title("Study Hours vs Marks")
axes[1,1].set_xlabel("Study Hours")
axes[1,1].set_ylabel("Marks")


# Main title
fig.suptitle("Data Analysis Charts", fontsize=16, fontweight="bold")


# Adjust spacing
plt.tight_layout()

# Display
plt.show()
```

###### Q13. Area Chart
###### Create an area chart:
###### months = ["Jan", "Feb", "Mar", "Apr", "May"]
###### revenue = [100, 150, 120, 180, 200]
###### Requirements:
###### Line on top
###### Filled area below the line
###### Title
###### X/Y labels
###### Focus: Remember:
###### fill_between()

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [100, 150, 120, 180, 200]
fig,ax=plt.subplots(figsize=(8,6))
ax.plot(months,revenue,marker="o",markerfacecolor="red",markeredgecolor="black",label="revenue")
ax.fill_between(months,revenue,alpha=0.5,color="black")
ax.set_title("Sales Analysis")
ax.set_xlabel("Months")
ax.set_ylabel("Revenue")
ax.legend()
plt.show()
```

##### Q14. Date-Based Chart
##### Create a monthly sales chart using:
##### dates = [
#####     "2026-01-01",
#####     "2026-02-01",
#####     "2026-03-01",
#####     "2026-04-01"
##### ]
##### sales = [100, 150, 120, 180]
##### Convert dates to datetime and plot them.
##### Rotate X-axis labels by 45 degrees.
##### Focus:
##### pd.to_datetime() + plot() + tick_params().

```python
import pandas as pd
dates = [
"2026-01-01",
"2026-02-01",
"2026-03-01",
"2026-04-01"
]
sales = [100, 150, 120, 180]
dates = pd.to_datetime(dates,errors="coerce")
fig,ax=plt.subplots(figsize=(6,5))
ax.plot(dates,sales)
ax.tick_params(axis="x",rotation=45) # rotates the X-axis labels by 45°.
ax.set_xlabel("Date")
ax.set_ylabel("Sales")
ax.set_title("Monthly Sales")
plt.show()
```

###### Q15. 🔥 Final Level 1 + Level 2 Challenge
###### Create a Sales Analysis Dashboard using 2 × 2 subplots.
###### Use:
###### months = ["Jan", "Feb", "Mar", "Apr", "May"]
###### revenue = [100, 150, 130, 180, 220]
###### profit = [30, 50, 40, 60, 80]
###### products = ["Laptop", "Mobile", "Tablet", "Watch"]
###### product_sales = [50, 80, 40, 30]
###### Create:
###### Chart 1 — Revenue
###### Line chart with:
###### markers
###### title
###### grid
###### legend
###### Chart 2 — Profit
###### Bar chart with:
###### bar labels
###### title
###### Chart 3 — Revenue vs Profit
###### Two lines:
###### Revenue
###### Profit
###### legend
###### Chart 4 — Product Sales
###### Bar chart with:
###### bar labels
###### title
###### And finally:
###### Sales Analysis Dashboard
###### as the main figure title.
###### Use:
###### plt.tight_layout()
###### appropriately.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [100, 150, 130, 180, 220]
profit = [30, 50, 40, 60, 80]

products = ["Laptop", "Mobile", "Tablet", "Watch"]
product_sales = [50, 80, 40, 30]

# Create 2 × 2 subplots
fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# Chart 1 — Revenue
ax[0, 0].plot(months, revenue, marker="o", label="Revenue")
ax[0, 0].set_title("Monthly Revenue")
ax[0, 0].grid(True)
ax[0, 0].legend()

# Chart 2 — Profit
bars = ax[0, 1].bar(months, profit)
ax[0, 1].set_title("Monthly Profit")

ax[0, 1].bar_label(bars)

# Chart 3 — Revenue vs Profit
ax[1, 0].plot(months, revenue, marker="o", label="Revenue")
ax[1, 0].plot(months, profit, marker="o", label="Profit")
ax[1, 0].set_title("Revenue vs Profit")
ax[1, 0].legend()

# Chart 4 — Product Sales
bars = ax[1, 1].bar(products, product_sales)
ax[1, 1].set_title("Product Sales")

ax[1, 1].bar_label(bars)

# Main title
fig.suptitle("Sales Analysis Dashboard", fontsize=16)

# Adjust layout
plt.tight_layout()

plt.show()
```
