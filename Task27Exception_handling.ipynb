##### 23. Create a function that checks password length and raises a custom exception if length is less than 8.

```python
class PasswordLengthError(Exception):
    pass
def check(password):
    if len(password)<=8:
        raise PasswordLengthError("Password length must be 8 or grater than 8")
    return password
try:
    password=input("Enter a password=")
    print(check(password))
except PasswordLengthError as e:
    print("Error :",e)
        
```
