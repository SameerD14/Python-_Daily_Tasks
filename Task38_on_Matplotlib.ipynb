##### Matplotlib Questions
##### 1. Create a simple line plot using:
##### o X = [1, 2, 3, 4, 5]
##### o Y = [10, 20, 15, 25, 30]

```python
import matplotlib.pyplot as plt
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]
plt.plot(X,Y)
plt.show()
```

##### 4. Plot two lines on the same graph and add a legend to identify them.

```python
x=[1,2,4,6,7]
y1=[2,5,6,7,9]
y2=[2,3,8,6,13]
plt.plot(x,y1,label="Y1 label")
plt.plot(x,y2,label="Y2 label")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Two line graph")
plt.legend()
plt.show()
```

##### 5. Change the line color, line style, and marker style of a line plot.

```python
x=[1,2,4,6,7]
y1=[2,5,6,7,9]
y2=[2,3,8,6,13]
plt.plot(x,y1,label="Y1 label",color="r",linestyle="-",marker="*")
plt.plot(x,y2,label="Y2 label",color="pink",linestyle="-",marker="*")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Two line graph")
plt.legend()
plt.show()
```

##### 6. Create a line plot using NumPy values generated from np.arange().

```python
import numpy as np
a=np.arange(1,10)
b=a**2
plt.plot(a,b,label="square values",color="pink",linestyle="-",marker="*")
plt.title("Square graph")
plt.xlabel("y axis")
plt.ylabel("y axis")
plt.yticks(b)

plt.legend()
plt.show()
```

```python
import numpy as np
a=np.arange(1,10)
plt.plot(a,label="numbers",color="pink",linestyle="-",marker="*")
plt.title("straight line")
plt.xlabel("x axis")
plt.ylabel("y axis")

plt.legend()
plt.show()
```

###### 7. Create a bar chart showing the sales of 5 different products.

```python
product=["watch","mobile","tablet","earbus","laptop"]
sales=[80,40,20,30,15]
plt.bar(product,sales)
plt.show()
```

##### 8. Create a horizontal bar chart for the same product-sales data.

```python
product=["watch","mobile","tablet","earbus","laptop"]
sales=[80,40,20,30,15]
plt.barh(product,sales,label="sales ratio")
plt.title("Product sales")
plt.legend()
plt.show()
```

##### 9. Create a scatter plot between Age and Salary for 10 employees.

```python
age=[22,25,28,30,34,23,34,24,26,32]
salary=[20000,30000,45000,43000,32000,45000,56000,89000,54000,24000]
plt.scatter(age,salary,color="r",label="salary by age",marker=".",s=200)
plt.title("Salary")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.legend()
plt.show()
```

##### 10.Create a histogram showing the distribution of 100 randomly generated
##### values.

```python
a=np.random.randint(10,80,10)
plt.hist(a,label="value count",color="Pink")
plt.title("Value Counts")
plt.xlabel("Values")
plt.ylabel("frequency")
plt.legend()
plt.show()
```
