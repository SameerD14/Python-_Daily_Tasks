##### 7. Read a file line by line using a loop.

```python
with open("student1.txt","r") as file:
    for line in file:
        print(line.strip())
```

##### 8. Write a program to copy content from one file to another file.

```python
with open("student1.txt", "r") as source:
    data = source.read()

with open("student2.txt", "w") as destination:
    destination.write(data)

print("File copied successfully.")
```

##### 9. Check if a file exists or not before reading it.

```python
import os

filename = "student1.txt"

if os.path.exists(filename):
    with open(filename, "r") as file:
        print(file.read())
else:
    print("File does not exist.")
```

##### 10. Write a program to delete a file (use os module).

```python
import os

filename = "student.txt"


if os.path.exists(filename):
    os.remove(filename)
    print("File deleted successfully.")
else:
    print("File does not exist.")
```

##### 16. Write a program to merge two text files into a third file.

```python
try:
    with open("student1.txt", "r") as file1:
        data1=file1.read()
    with open("Students.txt", "r") as file2:
        data2=file2.read()
    with open("File3.txt", "w") as file3:
        file3.write(data1)
        file3.write("\n")
        file3.write(data2)
    print("Files merged successfully.")
except FileNotFoundError:
    print("Error: One or both files do not exist.")

except Exception as e:
    print("Error:", e)
```

##### 17. Store a list of numbers in a file, then read and calculate their sum.

```python
numbers = [10, 20, 30, 40, 50]
with open("numbers.txt", "w") as file:
    for num in numbers:
        file.write(str(num) + "\n")
total = 0
with open("numbers.txt", "r") as file:
    for line in file:
        total += int(line)
print("Sum =", total)
```

##### 18. Create a file and write 10 lines. Then read only the first 5 lines.
