##### 20.Detect outliers in income_usd using the IQR method with NumPy/Pandas.

```python
import pandas as pd
import numpy as np

# Calculate Q1
Q1 = df["income_usd"].quantile(0.25)

# Calculate Q3
Q3 = df["income_usd"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = df[
    (df["income_usd"] < lower_bound) |
    (df["income_usd"] > upper_bound)
]

# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of outliers:", len(outliers))

print("\nOutliers:")
print(outliers[["income_usd"]])
```

##### 29.Create a final cleaned DataFrame containing:
##### • cleaned column names
##### • no duplicate IDs
##### • standardized categorical values
##### • numeric columns converted to numeric types
##### • invalid values converted to NaN
##### • missing values handled
##### • detected outliers handled

```python
import pandas as pd
import numpy as np

# Create final cleaned DataFrame
df_clean = df.copy()


# 1. Clean column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)


# 2. Remove duplicate IDs
df_clean = df_clean.drop_duplicates(subset="customer_id")


# 3. Convert numerical columns to numeric
numeric_cols = [
    "age",
    "income_usd",
    "credit_score",
    "tenure_years",
    "monthly_visits",
    "satisfaction"
]

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")


# 4. Convert invalid values to NaN
df_clean[numeric_cols] = df_clean[numeric_cols].replace(
    [-999, 99999],
    np.nan
)


# 5. Clean categorical columns
categorical_cols = [
    "customer_segment",
    "gender",
    "is_active"
]

for col in categorical_cols:
    df_clean[col] = df_clean[col].str.strip().str.lower()


# 6. Clean lead_source
df_clean["lead_source"] = (
    df_clean["lead_source"]
    .str.strip()
    .str.lower()
)

df_clean["lead_source"] = df_clean["lead_source"].replace({
    "website": "Website",
    "web": "Website",
    "referral": "Referral",
    "google ads": "Google Ads",
    "google_ads": "Google Ads"
})


# 7. Convert signup_date to datetime
df_clean["signup_date"] = pd.to_datetime(
    df_clean["signup_date"],
    errors="coerce",
    format="mixed"
)


# 8. Fill missing numerical values with median
df_clean[numeric_cols] = df_clean[numeric_cols].fillna(
    df_clean[numeric_cols].median()
)


# 9. Fill missing categorical values with mode
for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].mode()[0]
    )


# 10. Handle outliers using IQR
for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean[col] = df_clean[col].clip(lower, upper)


# Final cleaned DataFrame
print(df_clean.head())
```
