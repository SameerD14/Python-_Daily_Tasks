```python
# 1️⃣ Functions

# Q1. Write a function square() that takes a number and returns its square.

# Example:

# Input: 6
# Output: 36
def square(num):
    return num**2
print(square(10))
```

```python
# Q2. Write a function is_even() that takes a number and returns "Even" if it is even, otherwise "Odd".

def is_even(number):
    if number%2==0:
        return "Even"
    else:
        return "Odd"
print(is_even(34))
```

```python
# 2️⃣ List Comprehension

# Q3. Using list comprehension, create a list containing the squares of numbers from 1 to 10.

# Expected:

# [1, 4, 9, 16, ...]
print([i**2 for i in range(1,11)])
```

```python
# Q4. Given:

# numbers = [1, 2, 3, 4, 5, 6, 7, 8]

# Using list comprehension, create a list containing only the even numbers.
# Expected:

# [2, 4, 6, 8]
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
print([i for i in numbers if i%2==0])
```

```python
# 3️⃣ Lambda Function

# Q5. Write a lambda function that takes two numbers and returns their sum.

# Example:

# Input: 10, 20
# Output: 30
result=lambda x,y:x+y
print(result(2,4))

# Q6. Write a lambda function that takes a number and returns its cube.
result=lambda num:num**3
print(result(4))
```

##### 5. Bank Account
##### Attributes
##### • account_holder• balanceMethods• deposit()• withdraw()
##### Withdraw should happen only if balance is sufficient.

```python
class Bank_Account:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance
        
    def is_deposit(self,deposit):
        self.balance+=deposit
        print("Deposit is =",deposit)
        print("Current balance is =",self.balance)
    def is_withdraw(self,amount):
        print("Withdraw amontut is =",amount)
        if self.balance>amount:
            
            print("Balance is =",self.balance-amount)
        else:
            print("Insufficient Balance")
            print("Balance is =",self.balance-amount)
user1=Bank_Account("sameer Dhanjode",6000)
user1.is_deposit(8000)
user1.is_withdraw(7000)

```

##### 6. Car Speed Attributes • brand • speed
##### Method increase_speed(value) Maximum speed should never exceed 200 km/h.

```python
class Car:
    def __init__(self,brand,speed):
        self.brand=brand
        self.speed=speed
    def increase_speed(self,value):
        if self.speed+value<200:
            print("After increase speed is become=",self.speed+value,"km/hr")
        else:
            self.speed = 200
            print("Maximum speed is 200 km/h.")
            print("Current Speed =", self.speed, "km/h")
c1=Car("Lamborgini",120)
                  
print(c1.brand)
print(c1.speed)
c1.increase_speed(20)
c1.increase_speed(80)
```

##### 7. Temperature Store temperature. Method set_temperature() Condition Temperature should always remain between -20°C to 50°C Otherwise print Invalid.


```python
class Temperature:
    def __init__(self,temp):
        self.temp=temp
    def set_temperature(self):
        if -20<=self.temp<=50:
            print("Temperature=",self.temp)
        else:
            print("Invalid temperature")
        
t1=Temperature(90)
t2=Temperature(40)
t1.set_temperature()
t2.set_temperature()
```

```python
# 8. Movie Ticket Attributes • movie_name • seats
# Method book_ticket(number) Booking should happen only if seats are available.

class Movie:
    def __init__(self,movie_name,seats):
        self.movie_name=movie_name
        self.seats=seats
    def book_ticket(self,number):
        if self.seats>=number:
            print("Movie Name=",self.movie_name)
            print(f"Your {number} seats are booked")
            self.seats-=number
            print(f"Remainig seats are {self.seats}")
        else:
            print("Movie Name=",self.movie_name)

            print(f"seats are avialabel {self.seats}")
            print("Seats are not available:",number)
m1=Movie("Pushpa",200)
m1.book_ticket(160)
print("*"*20)
m1.book_ticket(60)
```
