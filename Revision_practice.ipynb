```python
# 1️⃣ Functions

# Q1. Write a function square() that takes a number and returns its square.

# Example:

# Input: 6
# Output: 36
def square(num):
    return num**2
print(square(10))
```

```python
# Q2. Write a function is_even() that takes a number and returns "Even" if it is even, otherwise "Odd".

def is_even(number):
    if number%2==0:
        return "Even"
    else:
        return "Odd"
print(is_even(34))
```

```python
# 2️⃣ List Comprehension

# Q3. Using list comprehension, create a list containing the squares of numbers from 1 to 10.

# Expected:

# [1, 4, 9, 16, ...]
print([i**2 for i in range(1,11)])
```

```python
# Q4. Given:

# numbers = [1, 2, 3, 4, 5, 6, 7, 8]

# Using list comprehension, create a list containing only the even numbers.
# Expected:

# [2, 4, 6, 8]
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
print([i for i in numbers if i%2==0])
```

```python
# 3️⃣ Lambda Function

# Q5. Write a lambda function that takes two numbers and returns their sum.

# Example:

# Input: 10, 20
# Output: 30
result=lambda x,y:x+y
print(result(2,4))

# Q6. Write a lambda function that takes a number and returns its cube.
result=lambda num:num**3
print(result(4))
```
