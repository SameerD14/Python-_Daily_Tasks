##### Basic Multiple Inheritance (1-10)
##### 1. Create two parent classes Father and Mother. Create a child class Child that inherits from
##### both.

```python
# Parent Class 1
class Father:
    def father_property(self):
        print("Father owns a House.")

# Parent Class 2
class Mother:
    def mother_property(self):
        print("Mother owns Gold.")

# Child Class
class Child(Father, Mother):
    def child_property(self):
        print("Child inherits from both Father and Mother.")

# Creating object
c = Child()

# Calling methods
c.father_property()
c.mother_property()
c.child_property()
```

##### 2. Define class A with method show() and class B with method display(). Create class C that
##### inherits from both and calls both methods.

```python
class A:
    def show(self):
        print("Class A")
class B:
    def display(self):
        print("Class B")
class C(A,B):
    def info(self):
        print("This is Child Class C")
c=C()
c.show()
c.display()
c.info()
    
```

##### 3. Create classes Teacher and Author. Create class Person that inherits from both and has
##### attributes from both.

```python
class Teacher:
    def work(self):
        print("Teacher teach student")
class Author:
    def write(self):
        print("Author write books")
class Person(Teacher,Author):
    def get(self):
        print("Person get knowledge from")
person=Person()
person.work()
person.write()
person.get()
```

##### 4. Define class Sports and class Academic. Create class Student that inherits from both.

```python
class Sports:
    def sports(self):
        print("Student plays Cricket")
class Academic:
    def study(self):
        print("Student studies Computer Science")
class Student(Sports, Academic):
    def student_info(self):
        print("Student Name : Sameer")
obj = Student()
obj.student_info()
obj.sports()
obj.study()
```

##### 5. Create class Camera and class MusicPlayer. Create class Smartphone that inherits from both.

```python
class Camera:
    def take(self):
        print("Take a picture")
class MusicPlayer:
    def play(self):
        print("Play a music")
class Smartphone(Camera,MusicPlayer):
    def name(self,name):
        self.name=name
        print("name=",self.name)
phone=Smartphone()
phone.name("Samsung")
phone.take()
phone.play()
```

##### 6. Define two classes with same method name. Create a child class that inherits from both and resolve the method conflict.

```python
class Father:
    def work(self):
        print("Father works")
class Mother:
    def work(self):
        print("Mother works")
class Child(Father,Mother):
    def work(self):
        print("I am Child")
        Father.work(self)
        Mother.work(self)
child=Child()
child.work()
```

##### 7. Create class Engine and class Wheels. Create class Car that inherits from both.

```python
class Engine:
    def start(self):
        print("Vehicle has Engine")
class Wheels:
    def have(self):
        print("Vehicle have wheels")
class Car(Engine,Wheels):
    def fuel(self):
        print("car use fuel")
car=Car()
car.fuel()
car.start()
car.have()
```

##### 8. Define class Writer and class Singer. Create class Artist that inherits from both.

```python
class Writer:
    def write(self):
        print("Writer write book")
class Singer:
    def sing(self):
        print("Singer sing a song")
class Artist(Writer,Singer):
    def draw(self):
        print("Artist draw a art")
a=Artist()
a.draw()
a.write()
a.sing()
```

##### 9. Create class Bank and class Insurance. Create class Customer that inherits from both.

```python
class Bank:
    def bank_details(self):
        print("Bank Name : SBI")
        print("Account Balance : ₹50,000")
class Insurance:
    def insurance_details(self):
        print("Insurance Company : LIC")
        print("Policy Amount : ₹10,00,000")
class Customer(Bank, Insurance):
    def customer_details(self):
        print("Customer Name : Sameer")
        print("Customer ID : C101")
customer = Customer()
customer.customer_details()
customer.bank_details()
customer.insurance_details()
```
