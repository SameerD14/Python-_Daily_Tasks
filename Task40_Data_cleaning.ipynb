##### Data Cleaning — 20 Questions
##### 1. Column Names
##### Display all column names. Identify columns having leading/trailing spaces and
##### rename all columns into a consistent snake_case format.

```python
df.columns=df.columns.str.lower().str.replace(" ","_")
```

##### 2. Duplicate Records
##### Find the number of duplicate rows. Remove the duplicate records and verify that no

```python
print("Dupliacates Rows=",df.duplicated().sum())
df=df.drop_duplicates()
print("Dupliacates Rows=",df.duplicated().sum())
```

##### 3. Missing Values

##### 4. Missing Categorical Values
##### Handle missing values in city, product_category, and payment_method using an
##### appropriate strategy.

```python
df["city"]=df["city"].fillna(df["city"].mode()[0])
df["payment_method"]=df["payment_method"].fillna(df["payment_method"].mode()[0])
df["product_category"]=df["product_category"].fillna(df["product_category"].mode()[0])
```

##### 5. Missing Numerical Values
##### Handle missing values in age, unit_price, discount_pct, customer_rating, and
##### revenue. Explain why you selected mean, median, or another method.

```python
list1=["age","unit_price","discount_%","customer_rating","revenue"]
for col in list1:
    df[col]=pd.to_numeric(df[col],errors="coerce")
```

```python
df["age"] = df["age"].fillna(df["age"].median())
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].median())
df["discount_%"] = df["discount_%"].fillna(df["discount_%"].median())
df["customer_rating"] = df["customer_rating"].fillna(df["customer_rating"].median())
df["revenue"] = df["revenue"].fillna(df["revenue"].median())
```

##### 6. Date Cleaning
##### Convert order_date into a proper datetime format. Identify invalid date values and
##### handle them appropriately.

```python
df["order_date"]=pd.to_datetime(df["order_date"],errors="coerce",format="mixed")
```

```python
print("Invalid dates:", df["order_date"].isna().sum())
df = df.dropna(subset=["order_date"])
```

##### 7. Age Cleaning
##### The age column contains values such as 25 years, 30 yrs, 45Y, and N/A. Extract the
##### numerical age and convert the column into a numeric datatype.

```python
# Convert age to numeric
df["age"] = pd.to_numeric(df["age"], errors="coerce")

```

###### 8. Age Outliers
###### Identify unrealistic ages such as 120, 150, and 200. Detect them using the IQR
###### method and handle them appropriately.

```python
import numpy as np
Q1=df["age"].quantile(0.25)
Q3=df["age"].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
print("Lower limit:", lower_bound)
print("Upper limit:", upper_bound)
outliers=df[(df["age"]<lower_bound)|(df["age"]>upper_bound)]
print("Age outliers:")
print(outliers)
df.loc[(df["age"] < lower_bound) | (df["age"] > upper_bound), "age"] = np.nan
```

```python
df["age"] = df["age"].fillna(df["age"].median())
#### IQR → Find unusual values → Replace them with NaN → Fill NaN with median
```

###### 9. Unit Price Cleaning
###### The unit_price column contains values such as ₹15000, $20000, 25000 INR, and 10k.
###### Convert all valid values into a single numeric format.

```python
# Convert to string
df["unit_price"] = df["unit_price"].astype(str)

# Remove currency symbols and INR
df["unit_price"] = df["unit_price"].str.replace("₹", "", regex=False)
df["unit_price"] = df["unit_price"].str.replace("$", "", regex=False)
df["unit_price"] = df["unit_price"].str.replace("INR", "", regex=False)

# Handle k values
df.loc[df["unit_price"].str.lower().str.endswith("k"), "unit_price"] = (
    df["unit_price"].str.lower().str.replace("k", "", regex=False).astype(float) * 1000
)

# Convert to numeric
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")
```

##### 10. Unit Price Outliers
##### Detect extreme unit_price values using the IQR method. Compare the number of
##### outliers before and after treatment.

```python
# Calculate Q1, Q3 and IQR
Q1 = df["unit_price"].quantile(0.25)
Q3 = df["unit_price"].quantile(0.75)

IQR = Q3 - Q1

# Calculate limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Count outliers before treatment
outliers_before = ((df["unit_price"] < lower) |
                   (df["unit_price"] > upper)).sum()

print("Outliers before treatment:", outliers_before)

# Replace outliers with NaN
df.loc[
    (df["unit_price"] < lower) |
    (df["unit_price"] > upper),
    "unit_price"
] = np.nan

# Fill outliers with median
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].median())

# Count outliers after treatment
outliers_after = ((df["unit_price"] < lower) |
                  (df["unit_price"] > upper)).sum()

print("Outliers after treatment:", outliers_after)
```

###### 11. Units Sold Cleaning
###### Clean values such as 5 units, 10 pcs, twenty, N/A, and 3 and convert the column into
###### a numeric datatype.

```python
# Convert to string
df["units_sold"] = df["units_sold"].astype(str)

# Replace text values
df["units_sold"] = df["units_sold"].str.lower()
df["units_sold"] = df["units_sold"].str.replace("units", "", regex=False)
df["units_sold"] = df["units_sold"].str.replace("pcs", "", regex=False)

# Convert word to number
df["units_sold"] = df["units_sold"].str.replace(
    "twenty", "20", regex=False
)

# Convert to numeric
df["units_sold"] = pd.to_numeric(
    df["units_sold"],
    errors="coerce"
)
```

```python
print(df["units_sold"].head())
print(df["units_sold"].dtype)
```

##### 12. Discount Cleaning
##### Clean values such as 10%, 20 percent, 5 %, none, and N/A. Convert the final
##### discount_pct column into numeric values between 0 and 100.

```python
# Convert to string
df.rename(columns={"discount": "discount_pct"}, inplace=True)
# Convert to lowercase
df["discount_pct"] = df["discount_pct"].astype(str).str.lower()
# Remove % and percent
df["discount_pct"] = df["discount_pct"].str.replace("%", "", regex=False)
df["discount_pct"] = df["discount_pct"].str.replace("percent", "", regex=False)

# Convert to numeric
df["discount_pct"] = pd.to_numeric(
    df["discount_pct"],
    errors="coerce"
)
```

##### 13. Discount Outliers
##### Identify invalid discount values such as 75, 90, and 120. Determine which values are
##### unrealistic and handle them.

```python
# Find invalid discount values
print(df[df["discount_pct"] > 100])

# Replace invalid values with NaN
df.loc[df["discount_pct"] > 100, "discount_pct"] = np.nan

# Fill with median
df["discount_pct"] = df["discount_pct"].fillna(
    df["discount_pct"].median()
)
```

###### 14. City Standardization
###### Standardize inconsistent city values such as:
###### o Mumbai
###### o mumbai
###### o MUMBAI
###### o Mumbai
###### o Bangalore
###### o bangalore
###### o Hyderbad
###### 
###### Create one consistent set of city names.

```python
# Remove extra spaces and convert to lowercase
df["city"] = df["city"].str.strip().str.lower()

# Correct spelling mistakes
df["city"] = df["city"].replace({
    "mumbai": "Mumbai",
    "bangalore": "Bangalore",
    "hyderbad": "Hyderabad"
})
```

###### 15. Product Category Cleaning
###### Standardize values such as Electronics, electronics, ELECTRONICS, Home and Kitchen,
###### home & kitchen, and Beauty.

```python
df.rename(columns={
    "customerid": "customer_id",
    "orderdate": "order_date",
    "productcategory": "product_category",
    "customersegment": "customer_segment",
    "saleschannel": "sales_channel",
    "unitssold": "units_sold",
    "unitprice": "unit_price",
    "customerrating": "customer_rating",
    "isactive": "is_active",
    "paymentmethod": "payment_method",
    "grosssales": "gross_sales",
    "discountamount": "discount_amount"
}, inplace=True)
```

```python
# Remove spaces and convert to lowercase
df["product_category"] = (
    df["product_category"]
    .str.strip()
    .str.lower()
)

# Standardize category names
df["product_category"] = df["product_category"].replace({
    "electronics": "Electronics",
    "home and kitchen": "Home and Kitchen",
    "home & kitchen": "Home and Kitchen",
    "beauty": "Beauty"
})
```

###### 16. Gender Cleaning
###### Standardize values such as Male, male, Male, FEMALE, female , M, and F into
###### consistent categories.


```python
# Remove spaces and convert to lowercase
df["gender"] = df["gender"].str.strip().str.lower()

# Standardize gender values
df["gender"] = df["gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
})
```

###### 17. Customer Segment Cleaning Identify invalid values such as Prem, Reg, VIP, and Unknown. 
###### Map valid variations to the correct categories and decide how to handle invalid categories.

```python
# Remove spaces and convert to lowercase
df["customer_segment"] = df["customer_segment"].str.strip().str.lower()

# Standardize valid variations
df["customer_segment"] = df["customer_segment"].replace({
    "prem": "Premium",
    "premium": "Premium",
    "reg": "Regular",
    "regular": "Regular",
    "vip": "VIP"
})

# Replace invalid categories with NaN
df.loc[
    ~df["customer_segment"].isin(["Premium", "Regular", "VIP"]),
    "customer_segment"
] = np.nan

df["customer_segment"] = df["customer_segment"].fillna(
    df["customer_segment"].mode()[0]
)
```

##### 18. Binary Column Cleaning
##### Clean is_active and returned, where values contain combinations of:
##### o 0/1
##### o Yes/No
##### o Y/N
##### o Active/Inactive
##### o Returned/Not Returned
##### Convert both columns into a consistent 0/1 binary format.

```python
# Clean is_active
df["is_active"] = df["is_active"].astype(str).str.strip().str.lower()

df["is_active"] = df["is_active"].replace({
    "1": 1,
    "yes": 1,
    "y": 1,
    "active": 1,
    "0": 0,
    "no": 0,
    "n": 0,
    "inactive": 0
})
```

```python
df["is_active"] = pd.to_numeric(df["is_active"], errors="coerce")
df["returned"] = pd.to_numeric(df["returned"], errors="coerce")
```
