###### Q1. Basic Line Chart
###### 
###### Create a line chart using:
###### 
###### months = ["Jan", "Feb", "Mar", "Apr", "May"]
###### sales = [100, 150, 120, 180, 200]
###### 
###### Requirements:
###### 
###### Title: Monthly Sales
###### X-axis: Month
###### Y-axis: Sales

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]
plt.plot(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

###### Q3. Sales Comparison
###### Create a line chart comparing:
###### months = ["Jan", "Feb", "Mar", "Apr", "May"]
###### sales_2025 = [100, 120, 140, 130, 160]
###### sales_2026 = [120, 150, 160, 170, 200]
###### Requirements:
###### Two lines
###### Markers
###### Different labels
###### Legend
###### Title
###### X/Y labels
###### Grid

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales_2025 = [100, 120, 140, 130, 160]
sales_2026 = [120, 150, 160, 170, 200]
plt.plot(months,sales_2025,marker="o",label="sales_2025",color="r",lw=2)
plt.plot(months,sales_2026,marker="s",label="sales_2026",color="green",lw=3)
plt.title("Sales Comparison")
plt.xlabel("Months")
plt.ylabel("sales")
plt.legend()
plt.grid()
plt.show()
```

##### Q8. Pie Chart

###### Q9. Subplots
###### Create 4 plots in one figure using 2 × 2 subplots:
###### Line chart
###### Bar chart
###### Histogram
###### Scatter plot
###### Use any of the datasets from previous questions.
###### Also give each subplot a title.
###### Hint:
###### fig, axes = plt.subplots(2, 2)

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]
fig,axes=plt.subplots(2,2)
axes[0,0].plot(months,sales,label="sales",lw=2,color="black",marker="^",ms=6,mfc="blue")
axes[0,0].set_title("Line Chart")
axes[0,0].legend()
axes[0,0].set_xlabel("Months")
axes[0,0].set_ylabel("Sales")
axes[0,1].bar(months,sales,label="sales",lw=2,color="blue")
axes[0,1].set_title("Sales")
axes[0,1].legend()
axes[0,1].set_xlabel("Months")
axes[0,1].set_ylabel("Sales")
axes[1,0].hist(sales,bins=5,label="sales",color="pink")
axes[1,0].set_title("Sales")
axes[1,0].legend()
axes[1,0].set_xlabel("Months")
axes[1,1].scatter(months,sales,label="sales",color="red",marker="^",s=50)
axes[1,1].set_title("Sales")
axes[1,1].legend()
axes[1,1].set_xlabel("Months")
axes[1,1].set_ylabel("Sales")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]

fig, axes = plt.subplots(2, 2)

# 1. Line Chart
axes[0, 0].plot(
    months,
    sales,
    label="Sales",
    lw=2,
    color="black",
    marker="^",
    ms=6,
    mfc="blue"
)

axes[0, 0].set_title("Line Chart")
axes[0, 0].legend()
axes[0, 0].set_xlabel("Months")
axes[0, 0].set_ylabel("Sales")


# 2. Bar Chart
axes[0, 1].bar(
    months,
    sales,
    label="Sales",
    color="blue"
)

axes[0, 1].set_title("Bar Chart")
axes[0, 1].legend()
axes[0, 1].set_xlabel("Months")
axes[0, 1].set_ylabel("Sales")


# 3. Histogram
axes[1, 0].hist(
    sales,
    bins=5,
    color="pink"
)

axes[1, 0].set_title("Histogram")
axes[1, 0].set_xlabel("Sales")
axes[1, 0].set_ylabel("Frequency")


# 4. Scatter Plot
axes[1, 1].scatter(
    months,
    sales,
    label="Sales",
    color="red",
    marker="^",
    s=50
)

axes[1, 1].set_title("Scatter Plot")
axes[1, 1].legend()
axes[1, 1].set_xlabel("Months")
axes[1, 1].set_ylabel("Sales")


plt.tight_layout()
plt.show()
```

##### Q10. Mini Data Analyst Task ⭐
##### Imagine you are a Data Analyst.
##### You have monthly revenue:
##### months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
##### revenue = [12000, 15000, 13500, 18000, 20000, 23000]
##### Create a professional-looking line chart.
##### Requirements:
##### Figure size (8, 5)
##### Line
##### Circle markers
##### Line width 2
##### Title
##### X-axis label
##### Y-axis label
##### Grid
##### Legend
##### Display the graph

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 15000, 13500, 18000, 20000, 23000]
plt.figsize(8,5)
plt.plot(months,revenue,marker="o",lw=2,label="revenue")
plt.title("Total Revenue")
plt.xlabel("Months")
plt.ylabel("Revenue")
plt.legend()
plt.grid()
plt.show()
```
