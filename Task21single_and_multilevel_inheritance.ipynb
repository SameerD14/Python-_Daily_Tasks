##### 1. Class Name: Person → Student
##### Problem
##### Create a parent class named Person and a child class named Student.
##### Explanation
##### The Person class should store:
##### • name
##### • age
##### The Student class should inherit from Person and add:
##### • marks
##### Create a method in Student to display the student's name, age, and marks.

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
class Student(Person):
    
    def __init__(self,name,age,mark):
        super().__init__(name,age)
        self.mark=mark
    def display(self):
        print("Student Name=",self.name)
        print("Age=",self.age)
        print("Mark=",self.mark)
std1=Student("Sameer",22,89)
std1.display()
```

##### 2. Class Name: Vehicle → Car
##### Problem
##### Create a parent class named Vehicle.
##### Explanation
##### The Vehicle class should contain a method:
##### • start()
##### Create a child class Car that:
##### • Adds a brand attribute.
##### • Overrides the start() method to display a car-specific starting message.

```python
class  Vehicle:
    def start(self):
        print("Vehicle is started ")
class Car(Vehicle):
    
    def __init__(self,brand):
        self.brand=brand
    def start(self):
        print(f"{self.brand} Car is started with push button")
car1=Car("Lamborgini")
car1.start()
    
```

##### 4. Class Name: Employee → Manager
##### Problem
##### Create a parent class named Employee.
##### Explanation
##### The Employee class should store:
##### • emp_id
##### • salary
##### Create child class Manager.
##### Add:
##### • department
##### Display all employee information.

```python
class Employee():
    def __init__(self,emp_id,salary):
        self.emp_id=emp_id
        self.salary=salary
        
class Manager(Employee):
    def __init__(self,emp_id,salary,department):
        super().__init__(emp_id,salary)
        self.department=department
    def  display_info(self):
        print("Manager Name=",self.emp_id)
        print("Salary=",self.salary)
        print("Department=",self.department)

emp1=Manager(18,60000,"Data Scientist")
emp1.display_info()
    
```

##### 5. Class Name: Shape → Circle

##### Explanation
##### Create a method:
##### • area()
##### Create child class:
##### • Circle
##### Store:
##### • radius
##### Override area() to calculate the area of a circle.

```python
class Shape():
    def calculate_area(self):
        print("Calculating area")
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    def calculate_area(self):
        print("Radius=",self.radius)
        print("Area=",3.14* (self.radius**2)) 
print("Circle")
cr1=Circle(4)
cr1.calculate_area()
```

##### 6. Class Name: BankAccount → SavingsAccount
##### Problem
##### Create a parent class named BankAccount.
##### Explanation
##### Provide methods:
##### • deposit()
##### • withdraw()
##### Create subclass:
##### • SavingsAccount
##### Add:
##### • calculate_interest()

```python
class BankAccount():
    def __init__(self,balance):
       self.balance=balance
    def deposit(self,amount):
        pass
    def withdraw(self,amount):
        pass
class Saving_account(BankAccount):
    def __init__(self,balance,interest_rate):
        super().__init__(balance)
        self.interest_rate=interest_rate
    def  deposit(self,amount):
        if amount>0:
            self.balance+=amount
            print("After deposit balance is ",self.balance)
        else:
            print("Invalid amount")
    def withdraw(self,amount):
        if amount<0:
            print("Invalid amount")
        elif amount>=self.balance:
            print("Insufficient Balance")
        else:
            self.balance-=amount
            print("After withdraw balance is ",self.balance)
    def calculate_interest(self):
        print("Interest Rate=",self.interest_rate,"%")
        print("After cut interest balance is",self.balance*(100-self.interest_rate)/100)
        
        
        
acc1=Saving_account(90000,10)
print("Balance is ",acc1.balance)
acc1.deposit(8000)
acc1.withdraw(18000)
print("Calculate Interest Rate")
acc1.calculate_interest()
    
    
```

##### 7. Class Name: Product → Electronics

##### Explanation
##### Store:

##### Create child class:
##### • Electronics
##### Add:

##### Display all information.

```python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def discount(self,discount):
        pass
class Eletronics(Product):
    def __init__(self,name,price):
        super().__init__(name,price)
    def discount(self,amount):
        self.price=self.price*(100-amount)/100
        print("After discount add price=",self.price)
    def warranty(self):
        print("Warranty of 3 years")
    def display(self):
        print("Product name=",self.name)
        print("Price=",self.price)

lap=Eletronics("Hp victus",90000)
lap.display()
lap.discount(10)
lap.warranty()
    
        
        
        
    
    
```

##### 8. Class Name: Book → TextBook
##### Problem
##### Create a parent class named Book.
##### Explanation
##### Store:
##### • title
##### • author
##### Create subclass:
##### • TextBook
##### Add:
##### • subject
##### Display complete details.

```python
class Named_Book:
    def __init__(self,title,author):
        self.title=title
        self.author=author
class Text_Book(Named_Book):
    def __init__(self,title,author,subject):
        super().__init__(title,author)
        self.subject=subject
    def display(self):
        print("Book Name=",self.title)
        print("Author name=",self.author)
        print("Subject=",self.subject)
book1=Text_Book("Discovery_of_India","Pandit Jawarlal Nehru","About India")
book1.display()
```
