##### 6. Class Name: BankAccount → SavingsAccount
##### Problem
##### Create a parent class named BankAccount.
##### Explanation
##### Provide methods:
##### • deposit()
##### • withdraw()
##### Create subclass:
##### • SavingsAccount
##### Add:
##### • calculate_interest()

```python
class BankAccount():
    def __init__(self,balance):
       self.balance=balance
    def deposit(self,amount):
        pass
    def withdraw(self,amount):
        pass
class Saving_account(BankAccount):
    def __init__(self,balance,interest_rate):
        super().__init__(balance)
        self.interest_rate=interest_rate
    def  deposit(self,amount):
        if amount>0:
            self.balance+=amount
            print("After deposit balance is ",self.balance)
        else:
            print("Invalid amount")
    def withdraw(self,amount):
        if amount<0:
            print("Invalid amount")
        elif amount>=self.balance:
            print("Insufficient Balance")
        else:
            self.balance-=amount
            print("After withdraw balance is ",self.balance)
    def calculate_interest(self):
        print("Interest Rate=",self.interest_rate,"%")
        print("After cut interest balance is",self.balance*(100-self.interest_rate)/100)
        
        
        
acc1=Saving_account(90000,10)
print("Balance is ",acc1.balance)
acc1.deposit(8000)
acc1.withdraw(18000)
print("Calculate Interest Rate")
acc1.calculate_interest()
    
    
```

##### 19. Class Name:
##### Shape → Polygon → Rectangle
##### Problem
##### Create multi-level inheritance.
##### Explanation
##### Rectangle should calculate its area using length and width.

```python
class Shape:
    def __init__(self):
        print("Shape class Constructor")
class Polygon(Shape):
    def __init__(self):
        super().__init__()
        print("Polygon class constructor")
class Rectraingle(Polygon):
    def __init__(self,length,width):
        super().__init__()
        self.length=length
        self.width=width
    def calculate_area(self):
        return self.length*self.width
rect=Rectraingle(12,8)
print("Length =", rect.length)
print("Width =", rect.width)
print("Area=",rect.calculate_area())
        
    
```

##### 20. Class Name:
##### BankAccount → SavingsAccount → RecurringDeposit
##### Problem
##### Create three levels of inheritance.
##### Explanation
##### Each class should add new functionality.
##### Example:
##### • deposit
##### • interest
##### • recurring deposit maturity

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print("Deposited Amount =", amount)
        print("Current Balance =", self.balance)
class SavingsAccount(BankAccount):
    def __init__(self, balance, interest_rate):
        super().__init__(balance)
        self.interest_rate = interest_rate
    def calculate_interest(self):
        interest = (self.balance * self.interest_rate) / 100
        print("Interest =", interest)
        return interest
class RecurringDeposit(SavingsAccount):
    def __init__(self, balance, interest_rate, monthly_deposit, months):
        super().__init__(balance, interest_rate)
        self.monthly_deposit = monthly_deposit
        self.months = months
    def maturity_amount(self):
        total = self.balance + (self.monthly_deposit * self.months)
        interest = (total * self.interest_rate) / 100
        maturity = total + interest
        print("Total Deposit =", total)
        print("Interest =", interest)
        print("Maturity Amount =", maturity)
rd = RecurringDeposit(10000, 8, 2000, 12)
rd.deposit(5000)
rd.calculate_interest()
rd.maturity_amount()
```
