##### 29. Class Name:
##### Shape → TwoDShape → Triangle
##### Problem
##### Create multi-level inheritance.
##### Explanation
##### Triangle should calculate its area using Heron's Formula.
##### Store:
##### • side1
##### • side2
##### • side3
##### Implement the area calculation.

```python
import math
class Shape:
    def __init__(self, name):
        self.name = name
class TwoDShape(Shape):
    def __init__(self, name):
        super().__init__(name)
class Triangle(TwoDShape):
    def __init__(self, name, side1, side2, side3):
        super().__init__(name)
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3
    def calculate_area(self):
        s = (self.side1 + self.side2 + self.side3) / 2
        area = math.sqrt(s * (s - self.side1) * (s - self.side2) * (s - self.side3))
        return area
    def display(self):
        print("Shape Name :", self.name)
        print("Side 1     :", self.side1)
        print("Side 2     :", self.side2)
        print("Side 3     :", self.side3)
        print("Area       :", self.calculate_area())
t = Triangle("Triangle", 3, 4, 5)
t.display()
```

##### 30. Class Name:
##### GrandParent → Parent → Child → GrandChild
##### Problem
##### Create four levels of inheritance.

##### Each class should have its own method.
##### The GrandChild object should be able to access methods from all parent classes.
##### Demonstrate Python's Method Resolution Order (MRO) using:
##### ClassName.mro()
##### or
##### help(ClassName)

```python
class GrandParent:
    def grandparent_method(self):
        print("This is GrandParent class method.")
class Parent(GrandParent):
    def parent_method(self):
        print("This is Parent class method.")
class Child(Parent):
    def child_method(self):
        print("This is Child class method.")
class GrandChild(Child):
    def grandchild_method(self):
        print("This is GrandChild class method.")


# Object Creation
obj = GrandChild()
obj.grandparent_method()
obj.parent_method()
obj.child_method()
obj.grandchild_method()
# Display Method Resolution Order (MRO) it tells how it arannge in memoryy
print("\nMethod Resolution Order (MRO):")
print(GrandChild.mro())
```
