##### 1. Take two numbers from user. Perform all arithmetic operations and also check
##### which number is greater using comparison operators.

```python
a=int(input("Enter a number="))
b=int(input("Enter a number2="))
print("Addition=",a+b)
print("Subtraction=",a-b)
print("Multiplication=",a*b)
print("Power=",a**b)
if b!=0:    
   print("Division=",a/b)
   print("Modulo=",a%b)
   print("Floor_Division=",a//b)
if a>b:
    print("Greater no=",a)
elif a<b:
    print("Greater no=",b)
else:
    print("Both no are equal")
```

##### 2. Print all numbers from 1 to 50. Skip multiples of 5 using continue and stop
##### completely when number 40 is reached using break.

```python
for i in range(1,51):
    if i%5==0:
        continue
    elif i>=40:
        break
    print(i)
    
```

##### 3. Take numbers from user until they enter negative number. Calculate sum and
##### average using while loop.

```python
total=0
count=0
while True:
    
    num=int(input("Enter a number="))
    if num<0:
        break
    total+=num
    count+=1
    avg=total/count
    
print("Addition=",total)
print("Average=",avg)
```

##### 4. Print the following pattern using nested loops:
##### *****
##### ****
##### ***
##### **
##### *

```python
for i in range(5,0,-1):
    for j in range(i):
        print("*",end=" ")
    print()
    
```

##### 5. Find the factorial of a number using for loop and also using while loop.

```python
num=int(input("Enter a number="))
i=1
factorial=1
while i<=num:
    factorial*=i
    i+=1
print("Factorial=",factorial)
```
