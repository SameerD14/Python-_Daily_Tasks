#####  create the class Employee take 3 arguments id ,salary and pf money. get the private attributes. salary>10000, pf money>1000. create the method to display the employee details and calculate the total salary. create the object of the class and call the methods

```python
class Employee:
    def __init__(self, id, salary, pf_money):
        self.id = id
        self.__salary = salary
        self.__pf_money = pf_money
    # Getter Methods
    def get_salary(self):
        return self.__salary
    def get_pf_money(self):
        return self.__pf_money
    # Setter Methods
    def set_salary(self, new_salary):
        if new_salary > 10000:
            self.__salary = new_salary
        else:
            print("Salary must be greater than 10000")
    def set_pf(self, new_pf):
        if new_pf > 1000:
            self.__pf_money = new_pf
        else:
            print("PF money must be greater than 1000")
    # Display Employee Details
    def display(self):
        print("Employee ID :", self.id)
        print("Salary      :", self.__salary)
        print("PF Money    :", self.__pf_money)
    # Calculate Total Salary
    def total_salary(self):
        total = self.__salary + self.__pf_money
        print("Total Salary =", total)
# Object Creation
emp = Employee(34, 100000, 5000)
# Display Details
emp.display()
# Total Salary
emp.total_salary()
# Update Salary
emp.set_salary(300000)
# Update PF
emp.set_pf(8000)
print("\nAfter Updating")
emp.display()
emp.total_salary()
```

```python
# food management 
class Food(ABC):
    #constructor
    def __init__(self,f_name,f_price):
        self.__Food_name=f_name
        self.__Food_price=f_price
    # getter
    def get_food_name(self):
        return self.__Food_name
    def get_food_price(self):
        return self.__Food_price
    # abstract method
    @abstractmethod
    def prepare(self):
        pass
    # abstract method
    @abstractmethod
    def deliver(self):
        pass
    def display_all_details(self):
        print(f"Food Name:{self.__Food_name}")
        print(f"Food Name:{self.__Food_price}")
        self.prepare()
        self.deliver()

# child class
class Pizza(Food):
    def prepare(self):
        print("Pizza is prepared")
    def deliver(self):
        print("Pizza is deliver")

class Burger(Food):
    def prepare(self):
        print("Burger is prepared")
    def deliver(self):
        print("Burger is deliver")

# create the class object
pizza=Pizza("Pizaa",500)
pizza.get_food_name()
pizza.get_food_price()
burger=Burger("Burger",900)
burger.get_food_name()
burger.get_food_price()
pizza.display_all_details()
print("*"*80)
burger.display_all_details()

```

```python
# talke the parent :company
# take two argument name,role
#take peivate
#make two abstract method 1 works ,2 salary
# create two child class worker ,manager ,diplaay all info


# company management 
class Company(ABC):
    #constructor
    def __init__(self,name,role):
        self.__name=name
        self.__role=role
    # getter
    def get_name(self):
        return self.__name
    def get_role(self):
        return self.__role
    # abstract method
    @abstractmethod
    def work(self):
        pass
    # abstract method
    @abstractmethod
    def salary(self):
        pass
    def display_all_details(self):
        print(f"Name:{self.__name}")
        print(f"Role:{self.__role}")
        self.work()
        self.salary()

# child class
class Worker(Company):
    def work(self):
        print("Worker ")
    def salary(self):
        print("salary=",30000)

class Manager(Company):
    def work(self):
        print("Manager")
    def salary(self):
        print("salary",78000)

# create the class object
worker1=Worker("Amit","data scientist")
worker1.get_name()
worker1.get_role()
manager1=Manager("Sameer","data scientist")
manager1.get_name()
manager1.get_role()
worker1.display_all_details()
print("*"*80)
manager1.display_all_details()

```

```python
from abc import ABC, abstractmethod

# Abstract Class
class Agent(ABC):

    def __init__(self, agent_name, api_key, role):
        self.agent_name = agent_name
        self.__api_key = api_key      # Private Attribute
        self.role = role

    # Getter
    def get_api_key(self):
        return self.__api_key

    # Setter
    def set_api_key(self, api_key):
        self.__api_key = api_key

    # Abstract Method
    @abstractmethod
    def perform_task(self):
        pass


# Child Class 1
class FlightBookingAgent(Agent):

    def __init__(self, agent_name, api_key, role, airline):
        super().__init__(agent_name, api_key, role)
        self.airline = airline

    def perform_task(self):
        print("\n----- Flight Booking Agent -----")
        print("Agent Name :", self.agent_name)
        print("Role :", self.role)
        print("Airline :", self.airline)
        print("Task : Flight Booked Successfully")


# Child Class 2
class HotelBookingAgent(Agent):

    def __init__(self, agent_name, api_key, role, hotel_name):
        super().__init__(agent_name, api_key, role)
        self.hotel_name = hotel_name

    def perform_task(self):
        print("\n----- Hotel Booking Agent -----")
        print("Agent Name :", self.agent_name)
        print("Role :", self.role)
        print("Hotel :", self.hotel_name)
        print("Task : Hotel Booked Successfully")


# Objects
flight = FlightBookingAgent("SkyAgent","API12345","Flight Booking","IndiGo")

hotel = HotelBookingAgent("StayAgent","API67890","Hotel Booking","Taj Hotel")

flight.perform_task()
hotel.perform_task()

print("\nPrivate API Key:", flight.get_api_key())

flight.set_api_key("NEWAPI999")
print("Updated API Key:", flight.get_api_key())
```

```python
# Grand Parent Class
class Person:
    def __init__(self, name, age):
        self.Name = name
        self.Age = age
    def get_name(self):
        return self.Name
    def get_age(self):
        return self.Age
    def set_age(self, age):
        if 20 <= age <= 60:
            self.Age = age
            print("Age updated successfully.")
        else:
            print("Age must be between 20 and 60.")
    def person_info(self):
        print("Name :", self.Name)
        print("Age  :", self.Age)
class Employee(Person):
    def __init__(self, name, age, salary, company):
        super().__init__(name, age)
        self.Employee_salary = salary
        self.Company_name = company
    def get_salary(self):
        return self.Employee_salary

    def get_company(self):
        return self.Company_name
    def set_salary(self, sal):
        if sal <= (20 / 100) * self.Employee_salary:
            self.Employee_salary += sal
            print("Salary updated successfully.")
        else:
            print("Increment should be less than or equal to 20%.")
    def employee_info(self):
        self.person_info()
        print("Company :", self.Company_name)
        print("Salary  :", self.Employee_salary)

class Manager(Employee):

    def __init__(self, name, age, salary, company, team_size, bonus):
        super().__init__(name, age, salary, company)
        self.Team_size = team_size
        self.Bonus = bonus

    def get_team_size(self):
        return self.Team_size

    def get_bonus(self):
        return self.Bonus
    def set_team(self, new_team):
        if 5 <= new_team<=10:
            self.Team_size = new_team
            print("Team size updated.")
        else:
            print("Team size must be between 5 and 10.")

    def set_bonus(self, new_bonus):
        if new_bonus <= (10 / 100) * self.Employee_salary:
            self.Bonus = new_bonus
            print("Bonus updated.")
        else:
            print("Bonus should not exceed 10% of salary.")

    # Display
    def display_all_info(self):
        self.employee_info()
        print("Team Size :", self.Team_size)
        print("Bonus     :", self.Bonus)
# Create Object
manager1 = Manager("Sameer", 25, 50000, "Microsoft", 8, 4000)

# Display Information
manager1.display_all_info()

print()

# Getters
print(manager1.get_name())
print(manager1.get_age())
print(manager1.get_salary())
print(manager1.get_company())
print(manager1.get_team_size())
print(manager1.get_bonus())

print()

# Setters
manager1.set_age(30)
manager1.set_salary(5000)
manager1.set_team(9)
manager1.set_bonus(3000)

print()

# Display Updated Information
manager1.display_all_info()
```

```python
class Count:

    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 5:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration
obj=Count()
for i in obj:
    print(i)
```

```python
import turtle
import math
import random

screen = turtle.Screen()
screen.bgcolor("black")

t = turtle.Turtle()
t.speed(0)
t.hideturtle()
t.pensize(1)

colors = [
    "red",
    "blue",
    "lime",
    "yellow",
    "cyan",
    "magenta",
    "orange",
    "pink"
]

for i in range(120):
    t.penup()
    t.goto(0, 40)

    angle = i * (math.pi * 2) / 120

    x = 16 * (math.sin(angle) ** 3)

    y = (
        13 * math.cos(angle)
        - 5 * math.cos(2 * angle)
        - 2 * math.cos(3 * angle)
        - math.cos(4 * angle)
    ) * 15

    c = random.choice(colors)
    t.color(c)

    t.pendown()
    t.goto(x, y)

    for _ in range(8):
        t.forward(6)
        t.backward(6)
        t.right(45)

turtle.done()
```
